We would also like some proxy for how much the borgerforslag has been shared online, as this is likely a large determinant of much much exposure it gets, thus how many signatures. We assume that Facebook is the most common platform for shares of this, and thus of largest effect of the social media. We estimate a proxy of how much a borgerforslag has been shared on Facebook by the following procedure. Firstly, remember that hard quotes in a google search searches for results that include the exact string.
- A google search is made, such that it includes the title of the borgerforslag in surrounded by quotes (") and specified facebook site by: `site:facebook.com`
So one search query could look like: 

`"Ret til specialiseret tværfaglig og koordineret behandling for personer med hjernerystelse" site:facebook.com`

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import re
import time

# Set Chrome options to disable the search engine choice screen
chrome_options = Options()
chrome_options.add_argument("--disable-search-engine-choice-screen")

# Initialize the Selenium Chrome driver with the specified options
driver = webdriver.Chrome(options=chrome_options)
driver.get("https://google.dk")
print("Opened Google.dk")

# Accept cookies if prompted
try:
    WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.ID, 'W0wltc'))).click()
    print("Accepted cookies")
except Exception:
    print("No cookie consent needed")

# List of search terms
search_list = ["Example1", "Example2"]
results_count_list = []

for term in search_list:
    try:
        print(f"Searching for: {term}")
        
        # Locate the search box and enter the search term
        search_box = WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.ID, 'APjFqb')))
        search_box.clear()  # Clear any previous search term
        search_box.send_keys(f'"{term}" site:facebook.com' + u'\ue007')
        
        # Wait for the "Værktøjer" button to be clickable and click it
        time.sleep(1)  # Short sleep to ensure stability
        WebDriverWait(driver, 20).until(EC.presence_of_element_located((By.ID, 'hdtb-tls'))).click()
        print(f"Clicked 'Værktøjer' for {term}")
        
        # Extract and process the search result stats
        result_stats = WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.ID, 'result-stats'))).text
        match = re.search(r'Ca\. ([\d,\.]+) resultater', result_stats)
        results_count_list.append(int(match.group(1).replace('.', '')) if match else 0)
        print(f"Found {results_count_list[-1]} results for {term}")
        
        # Add a brief pause between searches
        time.sleep(2)
        
    except Exception as e:
        print(f"Failed to retrieve results for {term}, appending 0.")
        results_count_list.append(0)

# Final output
print("Results count for each search term:", results_count_list)

```
